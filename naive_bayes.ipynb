Importing libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Loading Dataset

```python
df = pd.read_csv("/content/spam.csv", encoding="latin-1")
df.head()
```

```python
print(df.columns)
print(df.shape)
print(df.info())
```

Prepocessing

```python
df=df[["v1","v2"]]
print(df.head())
df.columns=["label","message"]
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
df.drop_duplicates(inplace=True)
print(df["label"].unique())
print(df["label"].value_counts())
```

having two columns one is target spam or not spam(ham) and another is message recivced

Encode the label column

```python
df["label"]=df["label"].map({"ham":0,"spam":1})
print(df.head())
```

Train Test split for do not cause data leakage

```python
from sklearn.model_selection import train_test_split
x=df["message"]
y=df["label"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape)
print(x_test.shape)
```

converting message into numbers using tfidf vectoriser

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
x_train_vec=tfidf.fit_transform(x_train)
x_test_vec=tfidf.transform(x_test)
print(x_train_vec.shape)
print(x_test_vec.shape)
```

Model traing and evaluting

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score,classification_report
nb=MultinomialNB()
nb.fit(x_train_vec,y_train)
y_pred=nb.predict(x_test_vec)
print(accuracy_score(y_train,nb.predict(x_train_vec)))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
